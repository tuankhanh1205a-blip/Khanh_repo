```python
def max_min(nums):
    if not nums:
        raise ValueError("Input list is empty")

    return max(nums), min(nums)
print(max_min([3, 1, 7, 2]))
```

```python
def three_smallest(nums):
    if len(nums) < 3:
        raise ValueError("List must contain at least 3 elements")

    return sorted(nums)[:3]
print(three_smallest([7, 2, 5, 1, 9, 3]))
```

```python
def second_largest_and_smallest(nums):
    distinct = set(nums)
    if len(distinct) < 2:
        raise ValueError("Need at least 2 distinct values")

    sorted_vals = sorted(distinct)
    return sorted_vals[-2], sorted_vals[1]
print(second_largest_and_smallest([5, 1, 9, 2, 5, 1]))
```

```python
from collections import Counter

def same_elements(a, b):
    if len(a) != len(b):
        return False
    return Counter(a) == Counter(b)
same_elements([1,2,3,4,5], [5,4,3,2,1])      
same_elements([1,2,2,3], [2,1,3,2])         
same_elements([1,2,2,3], [1,2,3,3])          
```

```python
def is_strictly_increasing(nums):
    for i in range(len(nums) - 1):
        if nums[i] >= nums[i + 1]:
            return False
    return True
is_strictly_increasing([])           
is_strictly_increasing([5])         
is_strictly_increasing([1,2,3,4])    
is_strictly_increasing([1,2,2,3])     
is_strictly_increasing([3,2,1])      
```

```python
def xth_largest_even(nums, x):
    evens = [n for n in nums if n % 2 == 0]

    if len(evens) < x:
        raise ValueError("Not enough even numbers")

    evens.sort(reverse=True)
    return evens[x - 1]
xth_largest_even([1, 2, 3, 4, 5], 1)   
xth_largest_even([1, 2, 3, 4, 5], 2)   
xth_largest_even([2, 4, 6, 8], 3)      
```

```python
def linear_search(nums, x):
    for i in range(len(nums)):
        if nums[i] == x:
            return i
    return -1
linear_search([1, 2, 3, 4, 5], 3)    
linear_search([1, 2, 2, 2, 5], 2)    
linear_search([1, 3, 5, 7], 4)       
```

```python
def binary_search(nums, x):
    left, right = 0, len(nums) - 1

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == x:
            return mid
        elif nums[mid] < x:
            left = mid + 1
        else:
            right = mid - 1

    return -1
binary_search([1, 3, 5, 7, 9], 5)  
binary_search([1, 3, 5, 7, 9], 6)   
binary_search([1, 2, 2, 2, 3], 2)   
```

```python
def bubble_sort(arr):
    a = arr[:]          
    n = len(a)

    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True
        if not swapped:   
            break

    return a
bubble_sort([5, 1, 4, 2, 8])

bubble_sort([1, 2, 3])

```

```python
def selection_sort(arr):
    a = arr[:]          # copy để không sửa mảng gốc
    n = len(a)

    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if a[j] < a[min_idx]:
                min_idx = j
        a[i], a[min_idx] = a[min_idx], a[i]

    return a
selection_sort([64, 25, 12, 22, 11])

selection_sort([1, 2, 3])
```

```python
def merge_sorted(a, b):
    i = j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    # thêm phần còn lại
    result.extend(a[i:])
    result.extend(b[j:])

    return result
merge_sorted([1, 3, 5], [2, 4, 6])

merge_sorted([1, 2, 3], [])

merge_sorted([], [4, 5])

merge_sorted([1, 2, 2], [2, 3])

```

```python
def merge_sorted(a, b):
    i = j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    result.extend(a[i:])
    result.extend(b[j:])
    return result


def merge_sort(nums):
    # base case
    if len(nums) <= 1:
        return nums[:]

    mid = len(nums) // 2

    left = merge_sort(nums[:mid])
    right = merge_sort(nums[mid:])

    return merge_sorted(left, right)
merge_sort([5, 2, 4, 6, 1, 3])

merge_sort([])

merge_sort([7])

merge_sort([3, 1, 2, 1])

```

```python
def quick_sort(nums):
    arr = nums[:]  # copy để không đổi list gốc

    def _quick_sort(arr, low, high):
        if low < high:
            p = partition(arr, low, high)
            _quick_sort(arr, low, p - 1)
            _quick_sort(arr, p + 1, high)

    _quick_sort(arr, 0, len(arr) - 1)
    return arr
print(quick_sort([5, 2, 4, 6, 1, 3]))  
print(quick_sort([3, 3, 1, 2]))       
print(quick_sort([]))                  


```
