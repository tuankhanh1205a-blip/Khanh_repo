```python
from collections import Counter

def min_remove_to_beautiful(arr):
    freq = Counter(arr)
    remove = 0

    for x, c in freq.items():
        if c > x:
            remove += c - x
        elif c < x:
            remove += c

    return remove
arr = [1, 2, 2]
print(min_remove_to_beautiful(arr))
```

```python
def majority_element(arr):
    candidate = None
    count = 0

    # Bước 1: tìm ứng viên
    for x in arr:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1

    # Bước 2: kiểm tra lại
    if candidate is not None and arr.count(candidate) > len(arr) // 2:
        print(candidate)
arr = [2, 2, 1, 2, 3, 2, 2]
majority_element(arr)
```

```python
def conditional_sort(arr):
    even = []
    odd = []

    for x in arr:
        if x % 2 == 0:
            even.append(x)
        else:
            odd.append(x)

    even.sort()          # chẵn tăng dần
    odd.sort(reverse=True)  # lẻ giảm dần

    return even + odd
arr = [5, 2, 8, 1, 3, 4, 6]
print(conditional_sort(arr))
```

```python
def merge_arrays(arr1, arr2):
    i = j = 0
    res = []

    # xác định tăng hay giảm
    ascending = True
    if len(arr1) >= 2 and arr1[0] > arr1[1]:
        ascending = False

    while i < len(arr1) and j < len(arr2):
        if ascending:
            if arr1[i] <= arr2[j]:
                res.append(arr1[i])
                i += 1
            else:
                res.append(arr2[j])
                j += 1
        else:
            if arr1[i] >= arr2[j]:
                res.append(arr1[i])
                i += 1
            else:
                res.append(arr2[j])
                j += 1

    # thêm phần còn lại
    while i < len(arr1):
        res.append(arr1[i])
        i += 1

    while j < len(arr2):
        res.append(arr2[j])
        j += 1

    return res
a = [1, 3, 5, 7]
b = [2, 4, 6]

print(merge_arrays(a, b))
```

```python
from collections import Counter

n = int(input())
names = []

for _ in range(n):
    full_name = input().strip()
    last_name = full_name.split()[-1]   # lấy TÊN
    names.append(last_name)

count = Counter(names)

max_freq = max(count.values())
candidates = [name for name in count if count[name] == max_freq]

print(min(candidates))
```

```python
s = input()

numbers = []
current = ""

for ch in s:
    if ch.isdigit():
        current += ch
    else:
        if current != "":
            numbers.append(int(current))
            current = ""

# nếu chuỗi kết thúc bằng số
if current != "":
    numbers.append(int(current))

if numbers:
    total = sum(numbers)
    average = total / len(numbers)
else:
    total = 0
    average = 0

print(total)
print(average)
```

```python
s = input().strip()

stack = []
pairs = {')': '(', '}': '{', ']': '['}

valid = True

for ch in s:
    if ch in '({[':
        stack.append(ch)
    elif ch in ')}]':
        if not stack or stack[-1] != pairs[ch]:
            valid = False
            break
        stack.pop()

if stack:
    valid = False

if valid:
    print("Valid")
else:
    print("Invalid")
```

```python
def is_composite(x):
    if x <= 3:
        return False
    for i in range(2, int(x ** 0.5) + 1):
        if x % i == 0:
            return True
    return False


N = int(input())

best_start = None
best_len = 0

cur_start = None
cur_len = 0

# duyệt từ N+1 đến 2N
for x in range(N + 1, 2 * N + 1):
    if is_composite(x):
        if cur_len == 0:
            cur_start = x
        cur_len += 1
    else:
        if cur_len > best_len:
            best_len = cur_len
            best_start = cur_start
        cur_len = 0

# kiểm tra đoạn cuối
if cur_len > best_len:
    best_len = cur_len
    best_start = cur_start

# in kết quả
if best_len > 0:
    for i in range(best_start, best_start + best_len):
        print(i, end=" ")

```

```python
arr = list(map(int, input().split()))
n = len(arr)

dp = [1] * n        # dp[i]: độ dài LIS kết thúc tại i
prev = [-1] * n    # để truy vết lại dãy

for i in range(n):
    for j in range(i):
        if arr[j] < arr[i] and dp[j] + 1 > dp[i]:
            dp[i] = dp[j] + 1
            prev[i] = j

# tìm vị trí kết thúc LIS dài nhất
max_len = max(dp)
idx = dp.index(max_len)

# truy vết dãy LIS
lis = []
while idx != -1:
    lis.append(arr[idx])
    idx = prev[idx]

lis.reverse()
print(*lis)
```
