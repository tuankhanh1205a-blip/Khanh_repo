```python
def fibonacci_to_file(a: int, b: int, n: int):
    fib = []

    for i in range(n + 1):
        if i == 0:
            fib.append(a)
        elif i == 1:
            fib.append(b)
        else:
            fib.append(fib[i-1] + fib[i-2])

    with open("output.txt", "w", encoding="utf-8") as f:
        for x in fib:
            f.write(str(x) + " ")
fibonacci_to_file(1,1,7)
```

```python
with open("students.csv", "w", encoding="utf-8") as f:
    f.write("""STT,Ho,Ten,Ngay sinh,Dai so,THCS4,THCS1
1,Truong,Vo,Ky,10/10/1999,8,10,9
2,Chu,Chi,Nhuoc,13/9/1999,10,10,10
""")
def grade10(path: str):
    with open(path, "r", encoding="utf-8") as f:
        lines = f.readlines()
        
    for line in lines[1:]:   # bỏ dòng tiêu đề
        data = line.strip().split(",")
        
        ho = data[1]
        ten = data[2]
        thcs4 = int(data[-2])  # cột THCS4
        
        if thcs4 == 10:
            print(ho, ten)
grade10("students.csv")
```

```python
with open("numbers.txt", "w") as f:
    f.write("3 5 1 9 2")
def find_max_min(path: str):
    try:
        with open(path, "r") as f:
            data = f.read().split()
            nums = list(map(int, data))
            print(max(nums), min(nums))
    except:
        print("Mission failed")
find_max_min("numbers.txt")
```

```python
with open("product.txt", "w") as f:
    f.write("""5
-10 -10 5 2 3
""")
def maximumProduct(path: str) -> int:
    with open(path, "r") as f:
        lines = f.readlines()
        
    nums = list(map(int, lines[1].split()))
    nums.sort()

    # 2 âm lớn nhất * số dương lớn nhất
    prod1 = nums[0] * nums[1] * nums[-1]
    # 3 số lớn nhất
    prod2 = nums[-1] * nums[-2] * nums[-3]

    return max(prod1, prod2)
print(maximumProduct("product.txt"))
```

```python
def max_rob(money):
    if not money:
        return 0
    if len(money) == 1:
        return money[0]

    dp = [0] * len(money)
    dp[0] = money[0]
    dp[1] = max(money[0], money[1])

    for i in range(2, len(money)):
        dp[i] = max(dp[i-1], dp[i-2] + money[i])

    return dp[-1]
print(max_rob([2,7,9,3,1]))
```

```python
with open("zero.txt", "w") as f:
    f.write("""7
0 1 0 3 12 0 5
""")
def moveZeroes(fileName: str) -> list:
    with open(fileName, "r") as f:
        lines = f.readlines()

    nums = list(map(int, lines[1].split()))

    non_zero = [x for x in nums if x != 0]
    zeros = [0] * (len(nums) - len(non_zero))

    return non_zero + zeros
print(moveZeroes("zero.txt"))
```

```python
with open("time.csv", "w") as f:
    f.write("""Lap,Time
1,143.23
2,134.12
3,140.65
""")
def averageTime(path: str) -> float:
    with open(path, "r") as f:
        lines = f.readlines()[1:]  # bỏ header

    times = []
    for line in lines:
        _, time = line.strip().split(",")
        times.append(float(time))

    avg = sum(times) / len(times)
    return round(avg, 2)
print(averageTime("time.csv"))
```

```python
with open("movies.txt", "w", encoding="utf-8") as f:
    f.write("""Inception|8.8|Mind-blowing sci-fi
Movie A|7.5|Good but boring
Interstellar|8.6|Epic space adventure
""")
def findMovies(path: str) -> list[str]:
    result = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            name, rating, desc = line.strip().split("|")
            if float(rating) > 8.0 and "boring" not in desc.lower():
                result.append(name)
    return result
print(findMovies("movies.txt"))
```

```python
with open("shapes.txt", "w") as f:
    f.write("""TYPE: SQUARE
SIDE: 5.3

TYPE: RECTANGLE
WIDTH: 5.0 HEIGHT: 16.0

TYPE: CIRCLE
RADIUS: 3.5
""")
import math

def calcPerimeter(path: str):
    with open(path, "r") as f:
        lines = [line.strip() for line in f if line.strip() != ""]

    i = 0
    while i < len(lines):
        if lines[i].startswith("TYPE:"):
            shape = lines[i].split(":")[1].strip()

            if shape == "SQUARE":
                side = float(lines[i+1].split(":")[1])
                p = 4 * side
                print(f"{p:.2f}")
                i += 2

            elif shape == "RECTANGLE":
                parts = lines[i+1].split()
                width = float(parts[1])
                height = float(parts[3])
                p = 2 * (width + height)
                print(f"{p:.2f}")
                i += 2

            elif shape == "CIRCLE":
                r = float(lines[i+1].split(":")[1])
                p = 2 * math.pi * r
                print(f"{p:.2f}")
                i += 2
        else:
            i += 1
calcPerimeter("shapes.txt")
```
