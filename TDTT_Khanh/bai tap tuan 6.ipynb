```python

nums = list(map(int, input().split()))
cumulative_sum = []
total = 0
for num in nums:
    total += num
    cumulative_sum.append(total)
print(cumulative_sum)
```

```python

nums = tuple(map(int, input().split()))
k = int(input())
if len(nums) > 0:
    k = k % len(nums)  # Xử lý trường hợp k > len(nums)
    result = nums[k:] + nums[:k]
else:
    result = nums
print(result)
```

```python

s = input()          
pairs = s.split()    

result = {}

for pair in pairs:
    key, value = pair.split(":")
    if key not in result:
        result[key] = []
    result[key].append(value)

print(result)
```

```python
nums = list(map(int, input().split()))

result = {'positives': 0, 'negatives': 0, 'zeros': 0}

for x in nums:
    if x > 0:
        result['positives'] += 1
    elif x < 0:
        result['negatives'] += 1
    else:
        result['zeros'] += 1

print(result)
```

```python
d1 = eval(input())
d2 = eval(input())
result = {}
for k in d1:
    result[k] = d1[k]
for k in d2:
    result[k] = result.get(k, 0) + d2[k]
for k in sorted(result):
    print(k, result[k])
```

```python
nums = list(map(int, input().split()))
k = int(input())

pairs = []

n = len(nums)
for i in range(n):
    for j in range(i+1, n):
        if nums[i] + nums[j] == k:
            pairs.append((nums[i], nums[j]))

pairs.sort()
print(pairs)
```

```python
t = tuple(map(int, input().split()))

even = []
odd = []

for x in t:
    if x % 2 == 0:
        even.append(x)
    else:
        odd.append(x)

print(tuple(even))
print(tuple(odd))
```

```python
nums = list(map(int, input().split()))
count = {}

for x in nums:
    count[x] = count.get(x, 0) + 1

max_freq = max(count.values())
candidates = [x for x in count if count[x] == max_freq]

print(min(candidates))
```
