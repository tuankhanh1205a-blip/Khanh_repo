```python
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
arr = [1, 3, 5, 7, 9, 11]
target = 7

idx = binary_search(arr, 0, len(arr) - 1, target)
print(idx)   
```

```python
def count_occurrences(arr, x) -> int:
    count = 0
    for value in arr:
        if value == x:
            count += 1
    return count
arr = [1, 2, 3, 2, 4, 2, 5]
x = 2

print(count_occurrences(arr, x))  
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if arr[j] > arr[j + 1]:
                # hoán đổi
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
a = [5, 1, 4, 2, 8]
print(bubble_sort(a))
```

```python
def most_frequent(arr):
    count = {}
    max_count = 0
    result = None

    for x in arr:
        count[x] = count.get(x, 0) + 1
        if count[x] > max_count:
            max_count = count[x]
            result = x

    return result
arr = [3, 1, 2, 3, 2, 1, 1]

print(most_frequent(arr))

```

```python
def count_pairs(arr, X):
    count = 0
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] == X:
                count += 1
    return count
arr = list(map(int, input().split()))
X = int(input())
print(count_pairs(arr, X))

```

```python
def length_of_LIS(nums):
    n = len(nums)
    dp = [1] * n

    for i in range(n):
        for j in range(i):
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)
```

```python
import bisect

def min_interval(intervals, queries):
    intervals.sort(key=lambda x: x[0])
    res = []

    for q in queries:
        ans = float('inf')
        for l, r in intervals:
            if l <= q <= r:
                ans = min(ans, r - l + 1)
        res.append(ans if ans != float('inf') else -1)

    return res
```

```python
def count_smaller(nums):
    n = len(nums)
    res = [0] * n

    for i in range(n):
        for j in range(i + 1, n):
            if nums[j] < nums[i]:
                res[i] += 1
    return res

```

```python
import bisect

def max_subarray_mod(nums, m):
    prefix = 0
    prefixes = [0]
    best = 0

    for x in nums:
        prefix = (prefix + x) % m
        idx = bisect.bisect_right(prefixes, prefix)
        if idx < len(prefixes):
            best = max(best, (prefix - prefixes[idx] + m) % m)
        else:
            best = max(best, prefix)

        bisect.insort(prefixes, prefix)

    return best
```

```python
from collections import Counter

def harmonious(arr):
    count = Counter(arr)
    res = 0
    for x in count:
        if x + 1 in count:
            res = max(res, count[x] + count[x + 1])
    return res
```
