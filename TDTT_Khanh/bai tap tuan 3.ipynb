```python
x = float(input("Nhập số thực: "))
phan_nguyen = int(x)
if x >= 0:
    tron_xuong = phan_nguyen
else:
    tron_xuong = phan_nguyen - 1 if x != phan_nguyen else phan_nguyen
if x >= 0:
    tron_len = phan_nguyen + 1 if x != phan_nguyen else phan_nguyen
else:
    tron_len = phan_nguyen
if x >= 0:
    gan_nhat = phan_nguyen + 1 if x - phan_nguyen >= 0.5 else phan_nguyen
else:
    gan_nhat = phan_nguyen - 1 if phan_nguyen - x > 0.5 else phan_nguyen

print(tron_len, tron_xuong, gan_nhat)
```
