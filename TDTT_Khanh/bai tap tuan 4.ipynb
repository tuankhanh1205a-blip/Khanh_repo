```python
def reverse_number(x):
    rev = 0
    while x > 0:
        rev = rev * 10 + x % 10
        x //= 10
    return rev
def is_palindrome(x):
    return x == reverse_number(x)
n = int(input())
steps = 0
while not is_palindrome(n):
    n = n + reverse_number(n)
    steps += 1
print(steps, n)
```

```python
n = int(input())
result = []
i = 1
while i * i <= n:
    sq = i * i
    temp = sq
    used = set()
    ok = True
    if temp == 0:
        used.add(0)
    else:
        while temp > 0:
            digit = temp % 10
            if digit in used:
                ok = False
                break
            used.add(digit)
            temp //= 10
    if ok:
        result.append(sq)
    i += 1
if len(result) == 0:
    print("(no number)")
else:
    print(" ".join(map(str, result)))
```

```python
def collatz_length(x):
    length = 1
    while x != 1:
        if x % 2 == 0:
            x //= 2
        else:
            x = 3 * x + 1
        length += 1
    return length
n = int(input())
max_len = 0
result_x = 1
for x in range(1, n + 1):
    l = collatz_length(x)
    if l > max_len:
        max_len = l
        result_x = x
print(result_x, max_len)
```

```python
n = int(input())
count = 0
i = 1
while i * i <= n:
    if n % i == 0:
        d1 = i
        d2 = n // i

        if d1 % 2 == 0:
            count += 1
        if d2 != d1 and d2 % 2 == 0:
            count += 1
    i += 1
print(count)
```

```python
def sum_proper_divisors(n):
    if n <= 1:
        return 0
    s = 1 
    i = 2
    while i * i <= n:
        if n % i == 0:
            s += i
            if i != n // i:
                s += n // i
        i += 1
    return s
a = int(input())
b = int(input())
if sum_proper_divisors(a) == b and sum_proper_divisors(b) == a:
    print("true")
else:
    print("false")
```
