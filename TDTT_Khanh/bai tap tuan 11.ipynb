```python
def unique_sorted_student_ids(list1, list2) -> list[str]:
    return sorted(set(list1) | set(list2))
list1 = ["A102", "A101", "A103"]
list2 = ["A103", "A104", "A101"]
print(unique_sorted_student_ids(list1, list2))

```

```python
def top3_frequent_names(names) -> list[str]:
    freq = {}
    for name in names:
        freq[name] = freq.get(name, 0) + 1
    sorted_names = sorted(freq.items(), key=lambda item: item[1], reverse=True)
    return [name for name, _ in sorted_names[:3]]
names = ["Alice", "Bob", "Alice", "Eve", "Bob", "Alice", "Eve"]
print(top3_frequent_names(names))

```

```python
def most_frequent_triple(triples):
    if not triples:
        raise ValueError("Input list is empty")

    freq = {}
    for t in triples:
        freq[t] = freq.get(t, 0) + 1

    max_count = -1
    result = None
    for t, c in freq.items():
        if c > max_count:
            max_count = c
            result = t

    return result
triples = [(1,2,3), (4,5,6), (1,2,3), (7,8,9), (1,2,3)]
print(most_frequent_triple(triples))
```

```python
def reconcile_users(registered, paid):
    reg_set = set(registered)
    paid_set = set(paid)

    paid_not_registered = sorted(paid_set - reg_set)
    registered_not_paid = sorted(reg_set - paid_set)
    both = sorted(reg_set & paid_set)

    return paid_not_registered, registered_not_paid, both
registered = [1, 2, 3, 4, 4]
paid = [3, 4, 5, 6]

print(reconcile_users(registered, paid))
```

```python
def contains_duplicate(nums):
    seen = set()
    for x in nums:
        if x in seen:
            return True
        seen.add(x)
    return False
print(contains_duplicate([1, 2, 3, 4]))     
print(contains_duplicate([1, 2, 3, 1]))     
```

```python
def squares_div_by3_not6(arr):
    return [x*x for x in arr if x % 3 == 0 and x % 6 != 0]
arr = [1, 3, 6, 9, 12, 15]
print(squares_div_by3_not6(arr))
```

```python
def even_numbers_in_matrix(matrix):
    return [x for row in matrix for x in row if x % 2 == 0]
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8]
]

print(even_numbers_in_matrix(matrix))
```

```python
def flatten_2d(matrix):
    return [x for row in matrix for x in row]
matrix = [
    [1, 2, 3],
    [4, 5],
    [6]
]

print(flatten_2d(matrix))
```

```python
def contains_nearby_duplicate(nums, k):
    seen = set()
    for i, x in enumerate(nums):
        if x in seen:
            return True

        seen.add(x)

        if len(seen) > k:
            seen.remove(nums[i - k])

    return False
nums = [1, 2, 3, 1]
k = 3
print(contains_nearby_duplicate(nums, k))
```

```python
def merge_similar_items(items1, items2):
    weight_map = {}

    for value, weight in items1:
        weight_map[value] = weight_map.get(value, 0) + weight

    for value, weight in items2:
        weight_map[value] = weight_map.get(value, 0) + weight

    return [[value, weight_map[value]] for value in sorted(weight_map)]
items1 = [[1, 1], [4, 5], [3, 8]]
items2 = [[3, 1], [1, 5]]

print(merge_similar_items(items1, items2))
```

```python
def farm_configurations(n):
    result = []

    if n % 2 != 0:
        return result

    for cows in range(n // 4 + 1):
        remaining_legs = n - 4 * cows
        chickens = remaining_legs // 2

        if chickens >= 0:
            result.append((chickens, cows))

    return result
print(farm_configurations(10))
```

```python
def can_make_odd_sum(a):
    return any(x % 2 == 1 for x in a)
print(can_make_odd_sum([2, 4, 6]))        
print(can_make_odd_sum([2, 4, 5, 6]))     
print(can_make_odd_sum([1]))              
print(can_make_odd_sum([0, 0, 0, 1]))     
```
