```python
def merge_sorted_lists(a, b):
    i = j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    # thêm phần còn lại
    result.extend(a[i:])
    result.extend(b[j:])

    return result
a = [1, 3, 5, 7]
b = [2, 4, 6]
print(merge_sorted_lists(a, b))

```

```python
def count_less_equal(matrix, mid):
    n = len(matrix)
    m = len(matrix[0])
    i = n - 1
    j = 0
    count = 0

    while i >= 0 and j < m:
        if matrix[i][j] <= mid:
            count += i + 1
            j += 1
        else:
            i -= 1
    return count


# ===== INPUT =====
n, m = map(int, input().split())
matrix = [list(map(int, input().split())) for _ in range(n)]
k = int(input())

# đổi bài toán: lớn thứ k → nhỏ thứ t
t = n * m - k + 1

low = matrix[0][0]
high = matrix[n - 1][m - 1]

ans = low

while low <= high:
    mid = (low + high) // 2
    if count_less_equal(matrix, mid) >= t:
        ans = mid
        high = mid - 1
    else:
        low = mid + 1

print(ans)
```

```python
s = list(input().strip())
n = len(s)

done = False

for i in range(n):
    min_digit = '9'
    pos = -1

    for j in range(n - 1, i, -1):
        if s[j] < min_digit:
            min_digit = s[j]
            pos = j

    if pos != -1 and min_digit < s[i]:
        # không để chữ số đầu thành 0
        if i == 0 and min_digit == '0':
            continue
        s[i], s[pos] = s[pos], s[i]
        done = True
        break

# nếu không thể làm nhỏ hơn, vẫn phải đổi đúng 2 chữ số
if not done:
    s[-1], s[-2] = s[-2], s[-1]

print("".join(s))
```

```python
def can_transform(a, b):
    n, m = len(a), len(b)
    dp = [[False] * (m + 1) for _ in range(n + 1)]
    dp[0][0] = True

    for i in range(1, n + 1):
        if a[i - 1].islower():
            dp[i][0] = dp[i - 1][0]

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            if a[i - 1].upper() == b[j - 1]:
                if a[i - 1].isupper():
                    dp[i][j] = dp[i - 1][j - 1]
                else:
                    dp[i][j] = dp[i - 1][j - 1] or dp[i - 1][j]
            else:
                if a[i - 1].islower():
                    dp[i][j] = dp[i - 1][j]

    return dp[n][m]


# ===== MAIN =====
Q = int(input())
for _ in range(Q):
    a = input().strip()
    b = input().strip()
    print("YES" if can_transform(a, b) else "NO")
```

```python
def is_subarray(list1, list2):
    n, m = len(list1), len(list2)

    if n > m:
        return False

    for i in range(m - n + 1):
        if list2[i:i+n] == list1:
            return True
    return False
a = [2, 3]
b = [1, 2, 3, 4]
print(is_subarray(a, b))
```

```python
n = int(input())
a = list(map(int, input().split()))

# Bước 1: tìm i
i = n - 2
while i >= 0 and a[i] >= a[i + 1]:
    i -= 1

if i == -1:
    # hoán vị cuối → in hoán vị đầu
    a.reverse()
else:
    # Bước 2: tìm j
    j = n - 1
    while a[j] <= a[i]:
        j -= 1

    # Bước 3: swap
    a[i], a[j] = a[j], a[i]

    # Bước 4: reverse suffix
    a[i + 1:] = reversed(a[i + 1:])

print(*a)
```

```python
def canPlaceFlowers(flowerbed, k):
    n = len(flowerbed)

    for i in range(n):
        if flowerbed[i] == 0:
            left = (i == 0) or (flowerbed[i - 1] == 0)
            right = (i == n - 1) or (flowerbed[i + 1] == 0)

            if left and right:
                flowerbed[i] = 1
                k -= 1
                if k == 0:
                    return True

    return k <= 0
flowerbed = [1,0,0,0,1]
k = 1
print(canPlaceFlowers(flowerbed, k))
```

```python
def is_path_crossing(moves):
    x, y = 0, 0
    visited = {(0, 0)}

    for move in moves:
        if move == 'R':
            x += 1
        elif move == 'L':
            x -= 1
        elif move == 'U':
            y += 1
        elif move == 'D':
            y -= 1

        if (x, y) in visited:
            return True

        visited.add((x, y))

    return False
moves = "URDL"
print(is_path_crossing(moves))
```

```python
def Median(nums):
    a, b, c, d, e = nums

    # 1–2: sắp xếp từng cặp
    if a > b:
        a, b = b, a
    if c > d:
        c, d = d, c

    # 3: loại min nhỏ nhất
    if a > c:
        a, c = c, a
        b, d = d, b

    # 4: loại max lớn nhất
    if b > d:
        b = d

    # b, c, e là 3 ứng viên trung vị
    # 5–6: median của 3 số
    if b > c:
        b, c = c, b
    if b > e:
        return b
    elif c > e:
        return e
    else:
        return c
print(Median([5, 1, 9, 3, 7]))  
print(Median([1, 2, 3, 4, 5]))  
print(Median([9, 8, 7, 6, 5]))  
```
