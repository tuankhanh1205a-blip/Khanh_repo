```python
def max_two(a, b):
    if a > b:
        return a
    else:
        return b

a = int(input())
b = int(input())
print(max_two(a, b))
```

```python
def swap(a, b):
    return b, a
a = int(input())
b = int(input())
print(swap(a, b))
```

```python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True
n = int(input())
print(is_prime(n))
```

```python
def find_first_position(arr, k):
    for i in range(len(arr)):
        if arr[i] == k:
            return i
    return -1
arr = list(map(int,input().split()))
k = int(input())
print(find_first_position(arr, k))
```

```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
n = int(input())
print(factorial(n))
```

```python
def calculator(num1, num2, operat):
    if operat == '+':
        result = num1 + num2
    elif operat == '-':
        result = num1 - num2
    elif operat == '*':
        result = num1 * num2
    elif operat == '/':
        if num2 == 0:
            return None
        result = num1 / num2
    else:
        return None

    return round(result, 2)
print(calculator(5, 3, '+'))  
print(calculator(5, 3, '/'))  
```

```python
def hamming_distance(x, y):
    xor = x ^ y
    count = 0
    while xor > 0:
        count += xor & 1
        xor >>= 1
    return count
print(hamming_distance(1, 4))  
```

```python
def sum_digits(n):
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total
```
