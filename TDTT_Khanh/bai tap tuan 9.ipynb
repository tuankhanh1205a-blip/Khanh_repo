```python
class Cylinder:
    def __init__(self, r, h):
        self.r = r
        self.h = h

    def surface_area(self):
        return 2 * 3.14 * self.r * (self.r + self.h)

    def volume(self):
        return 3.14 * self.r * self.r * self.h


r, h = map(float, input().split())

cyl = Cylinder(r, h)

print(f"{cyl.surface_area():.2f} {cyl.volume():.2f}")
```

```python
class Date:
    def __init__(self, d, m, y):
        self.d = d
        self.m = m
        self.y = y

    def is_leap_year(self):
        return self.y % 400 == 0 or (self.y % 4 == 0 and self.y % 100 != 0)

    def is_valid(self):
        if self.y <= 0 or self.m < 1 or self.m > 12:
            return False

        days_in_month = [31, 29 if self.is_leap_year() else 28,
                         31, 30, 31, 30,
                         31, 31, 30, 31, 30, 31]

        if self.d < 1 or self.d > days_in_month[self.m - 1]:
            return False

        return True

    def next_day(self):
        days_in_month = [31, 29 if self.is_leap_year() else 28,
                         31, 30, 31, 30,
                         31, 31, 30, 31, 30, 31]

        self.d += 1
        if self.d > days_in_month[self.m - 1]:
            self.d = 1
            self.m += 1
            if self.m > 12:
                self.m = 1
                self.y += 1

    def format(self):
        return f"{self.d:02d}/{self.m:02d}/{self.y}"


# ===== MAIN =====
s = input().strip()
d, m, y = map(int, s.split('/'))

date = Date(d, m, y)

if not date.is_valid():
    print("INVALID")
else:
    date.next_day()
    print(date.format())
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def is_origin(self):
        return self.x == 0 and self.y == 0

    def is_on_y_axis(self):
        return self.x == 0 and self.y != 0

    def is_on_x_axis(self):
        return self.y == 0 and self.x != 0

    def distance_to(self, other):
        return math.sqrt((self.x - other.x) ** 2 + (self.y - other.y) ** 2)


# ===== MAIN =====
x, y = map(float, input().split())
M = Point(x, y)
O = Point(0, 0)

# kiểm tra vị trí
if M.is_origin():
    print("M la goc toa do")
elif M.is_on_y_axis():
    print("M nam tren truc tung")
elif M.is_on_x_axis():
    print("M nam tren truc hoanh")
else:
    print("M khong nam tren truc toa do")

# khoảng cách tới gốc tọa độ
dist = M.distance_to(O)
print(f"{dist:.2f}")
```

```python
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def set_numbers(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    def subtract(self):
        return self.a - self.b

    def product(self):
        return self.a * self.b

    def divide(self):
        if self.b == 0:
            return "Khong the chia cho 0"
        return self.a / self.b

    def power(self):
        return self.a ** self.b

    def mod(self):
        if self.b == 0:
            return "Khong the chia du cho 0"
        return self.a % self.b


# ===== MAIN =====
a, b = map(float, input("Nhap 2 so ban dau: ").split())
calc = Calculator(a, b)

while True:
    print("\n===== MENU =====")
    print("1. Cong")
    print("2. Tru")
    print("3. Nhan")
    print("4. Chia")
    print("5. Luy thua")
    print("6. Chia du")
    print("7. Cap nhat 2 so moi")

    choice = input("Chon phep toan (1-7): ")

    if choice == '1':
        print("Ket qua:", calc.add())
    elif choice == '2':
        print("Ket qua:", calc.subtract())
    elif choice == '3':
        print("Ket qua:", calc.product())
    elif choice == '4':
        print("Ket qua:", calc.divide())
    elif choice == '5':
        print("Ket qua:", calc.power())
    elif choice == '6':
        print("Ket qua:", calc.mod())
    elif choice == '7':
        a, b = map(float, input("Nhap 2 so moi: ").split())
        calc.set_numbers(a, b)
        print("Da cap nhat 2 so moi")
    else:
        print("Lua chon khong hop le")

    exit_choice = input("Ban co muon thoat? ")
    if exit_choice in ["yes", "YES", "Yes"]:
        break
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = []   # danh sách rỗng ban đầu

    def add_item(self, name, price):
        self.items.append({"name": name, "price": price})

    def remove_item(self, name):
        for item in self.items:
            if item["name"] == name:
                self.items.remove(item)
                return True
        return False

    def is_empty(self):
        return len(self.items) == 0

    def total_price(self):
        total = 0
        for item in self.items:
            total += item["price"]
        return total

    def show_items(self):
        if self.is_empty():
            print("Gio hang rong")
        else:
            for item in self.items:
                print(f"Ten: {item['name']}, Gia: {item['price']}")

    def clear_cart(self):
        self.items.clear()
cart = ShoppingCart()

while True:
    print("\n===== MENU =====")
    print("2. Them san pham")
    print("3. Xoa san pham theo ten")
    print("4. Kiem tra gio hang rong")
    print("5. Tinh tong tien")
    print("6. Hien thi gio hang")
    print("7. Xoa toan bo gio hang")

    choice = input("Nhap lua chon (2-7): ")

    if choice == "2":
        name = input("Nhap ten san pham: ")
        price = float(input("Nhap gia san pham: "))
        cart.add_item(name, price)
        print("Da them san pham")

    elif choice == "3":
        name = input("Nhap ten san pham can xoa: ")
        if cart.remove_item(name):
            print("Da xoa san pham")
        else:
            print("Khong tim thay san pham")

    elif choice == "4":
        if cart.is_empty():
            print("Gio hang rong")
        else:
            print("Gio hang khong rong")

    elif choice == "5":
        print("Tong tien:", cart.total_price())

    elif choice == "6":
        cart.show_items()

    elif choice == "7":
        cart.clear_cart()
        print("Da xoa toan bo gio hang")

    else:
        print("Lua chon khong hop le")

    exit_choice = input("Co muon thoat khoi chuong trinh khong? ")
    if exit_choice == "yes":
        break

```

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

    def perimeter(self):
        return 2 * (self.w + self.h)

    def scale(self, k):
        self.w *= k
        self.h *= k


# ===== MAIN =====
w, h, k = map(float, input().split())

rect = Rectangle(w, h)
rect.scale(k)

print(f"{rect.area():.2f} {rect.perimeter():.2f}")
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, x):
        self.balance += x

    def withdraw(self, x):
        if x <= self.balance:
            self.balance -= x

    def __str__(self):
        return f"{self.owner}: {self.balance:.2f}"


# ===== MAIN =====
owner, balance = input().split()
balance = float(balance)

account = BankAccount(owner, balance)

n = int(input())

for _ in range(n):
    cmd, x = input().split()
    x = float(x)
    if cmd == "DEPOSIT":
        account.deposit(x)
    elif cmd == "WITHDRAW":
        account.withdraw(x)

print(f"{account.balance:.2f}")
```

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year


class Library:
    def __init__(self):
        self.books = []

    def add(self, book):
        self.books.append(book)

    def count_by_author(self, name):
        count = 0
        for book in self.books:
            if book.author == name:
                count += 1
        return count

    def find_by_year(self, y):
        count = 0
        for book in self.books:
            if book.year == y:
                count += 1
        return count


# ===== MAIN =====
lib = Library()

n = int(input())
for _ in range(n):
    line = input().strip()

    if line.startswith("ADD"):
        _, data = line.split(" ", 1)
        title, author, year = data.split(";")
        year = int(year)
        book = Book(title, author, year)
        lib.add(book)

    elif line.startswith("COUNTYEAR"):
        _, year = line.split()
        year = int(year)
        print(lib.find_by_year(year))

    elif line.startswith("COUNT"):
        _, author = line.split(" ", 1)
        print(lib.count_by_author(author))

```
